```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.naive_bayes import 
```

```python
#initialization and data reading

pd.set_option('display.max_colwidth', None)
le = LabelEncoder()
df1 = pd.read_csv('spam_train1.csv')
df2 = pd.read_csv('spam_train2.csv')
```

Data Step

```python
# Removing unneeded columns for encoding

# df1.isna().sum(axis=0)
# df2.isna().sum(axis=0)

# labels and features
df2 = df2.iloc[:,1:-1]
df1 = df1.iloc[:,:-3]

# combining into one DF
df1.columns = df2.columns
df = pd.concat([df1, df2], axis=0)

#drop duplicates
df = df.drop_duplicates()
df.duplicated().sum() #sum 0
```

Data processing and encoding

```python
# Each sentence is turned into a list of words andthen put into their own row and numbered based on which email theyre from

df_split_text = df['text'].map(lambda x: x.split(' '))
dataframe_words = (
df_split_text.explode().str.lower()
 .str.replace(r'[^a-z]', '', regex=True)
 )
```

```python
# Each unique word is assigned two values: an id for the word and the email id for the email it came from 

dataframe_words_encoded = pd.DataFrame(le.fit_transform(dataframe_words))
dataframe_words = pd.DataFrame(dataframe_words)
dataframe_words = dataframe_words.reset_index()

df_encoded = pd.DataFrame(
    data=[
        dataframe_words.values[:,0],
        dataframe_words_encoded.iloc[:, 0].values,
        dataframe_words.values[:,1]
    ],
    index=["email_id", "word_id", "word"]
)
df_encoded = df_encoded.transpose()
```
