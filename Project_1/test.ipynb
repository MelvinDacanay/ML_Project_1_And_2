```python
df1 = pd.read_csv('spam_train1.csv')
df2 = pd.read_csv('spam_train2.csv')
```

Data Step

```python
# Removing unneeded columns for encoding

# df1.isna().sum(axis=0)
# df2.isna().sum(axis=0)

# labels and features
df2 = df2.iloc[:,1:-1]
df1 = df1.iloc[:,:-3]

# combining into one DF
df1.columns = df2.columns
df = pd.concat([df1, df2], axis=0)

#drop duplicates
df = df.drop_duplicates()
df.duplicated().sum() #sum 0
```

Data processing and encoding

```python
dfs = df['text'].map(lambda x: x.split(' '))
```

```python
dataframe_words = (
dfs.explode().str.lower()
 .str.replace(r'[^a-z]', '', regex=True)
 )
```

```python
df4b = pd.DataFrame(le.fit_transform(dataframe_words))
df4a = pd.DataFrame(dataframe_words)
```

```python
df4a = df4a.reset_index()
```
